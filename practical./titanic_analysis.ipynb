```python
missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing_data
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    df.isnull(),
    cmap="viridis",
    cbar=True,
    yticklabels=False
)

plt.title("Missing Value Heatmap")

plt.show()
```

```python
duplicate_rows = df[df.duplicated()]

duplicate_rows
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,8))

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
numerical_columns = df.select_dtypes(include=["int64","float64"]).columns

plt.figure(figsize=(14,8))

df[numerical_columns].boxplot()

plt.xticks(rotation=45)

plt.show()
```

```python
print("Rows with missing Age :", df["Age"].isnull().sum())
```

```python
survived = df[df["Survived"] == 1]

print("Passengers Survived :", len(survived))
```

```python
dead = df[df["Survived"] == 0]

print("Passengers Dead :", len(dead))
```

```python
df = df.dropna(subset=["Age"])

print(df.shape)
```

```python
df.loc[df["Survived"] == 1, ["Name", "Age"]]
```

```python
numeric_df = df.select_dtypes(include=np.number)

numeric_df
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Age"],
    kde=True,
    bins=20
)

plt.title("Age Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    x="Survived",
    data=df
)

plt.title("Count of Survived Passengers")

plt.show()
```

```python
survivors = df[df["Survived"] == 1]

print(survivors[["Name","Sex","Age"]])
```

```python
survivors["Sex"].value_counts()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=survivors,
    x="Sex"
)

plt.title("Male vs Female Survivors")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df[df["Survived"] == 1],
    x="Pclass"
)

plt.title("Survived Passengers by Passenger Class")

plt.show()
```
